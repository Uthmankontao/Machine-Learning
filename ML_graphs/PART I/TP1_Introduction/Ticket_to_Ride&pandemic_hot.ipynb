#### 8. Display the diameter, the radius and the girth of both graphs

```python
print(G.edges())
print(G.number_of_edges(4,5))
```

```python
def girth_multigraph(G):
    for u,v in G.edges():
        if G.number_of_edges(u,v) > 1:
            return 2
    G_cycle = nx.cycle_basis(G)
    return min(len(c) for c in G_cycle) if G_cycle else 'inf'    
```

```python
G_diam, G_ray = nx.diameter(G), nx.radius(G)
G_girth = girth_multigraph(G)
print(f"Le diamètre pour le graphe G: {G_diam}, son rayon: {G_ray} et son girth qui represente la longueur du plus petit cycle: {G_girth}")
```

```python
H_diam, H_ray = nx.diameter(H), nx.radius(H)
H_Hirth = girth_multigraph(H)
print(f"Le diamètre pour le Graphe H: {H_diam}, son rayon: {H_ray} et son girth qui represente la longueur du plus petit cycle: {H_Hirth}")
```
