#### 2. Determine the order and size of these graphs

L'ordre d'un graphe est le nombre de noeuds qui le composent et la taille désigne le nombre total d'arêtes entre chaque paire de noeuds.

```python
print(f"The order of ticket to ride's graph is {G.order()} and the size is {G.size()}")
print(f"The order of this pandemic's graph is {H.order()} and the size is {H.size()}")
```

#### 3. Show both graphs as adjacency lists

```python
G_adj_list = [(n, nbrdict) for n, nbrdict in G.adjacency()]
H_adj_list = [(n, nbrdict) for n, nbrdict in H.adjacency()]
print(G_adj_list)
print(H_adj_list)
```

#### 4. Determine and display the adjacency, incidence and Laplacian matrices of these graphs

```python
# matrice d'adjacence
G_Ma_adj = nx.adjacency_matrix(G, weight='weight').toarray()
# G_Ma_adj = G_Ma_adj.toarray()
print(G_Ma_adj)
```

```python
H_Ma_adj = nx.adjacency_matrix(H, weight='weight').toarray()
print(H_Ma_adj)
```

```python
# matrice d'incidence
G_Ma_inc = nx.incidence_matrix(G).toarray()
print(G_Ma_inc.shape)
G_Ma_inc[:5]
```

```python
H_Ma_inc = nx.incidence_matrix(H).toarray()
print(H_Ma_inc.shape)
H_Ma_inc[:5]
```

Matrice nxp où n est le nombre de noeuds et p le nombre d'arêtes. Si la case (i,j) = 1 cela veut dire que le sommet i est une extrémité de l'arête j, si c'est = 2 alors l'arête j est une boucle sur le sommet i et 0 sinon.

```python
# matrice laplacienne
G_Ma_lapl = nx.laplacian_matrix(G).toarray()
print(G_Ma_lapl)
```

```python
H_Ma_lapl = nx.laplacian_matrix(H).toarray()
print(H_Ma_lapl)
```
