```python
import torch
from torch.nn import Linear
from torch_geometric.nn import GCNConv # <--- Le truc spécifique

class MonGNN(torch.nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super().__init__()
        # On définit les couches (Message Passing)
        self.conv1 = GCNConv(input_dim, hidden_dim)
        self.conv2 = GCNConv(hidden_dim, output_dim)

    def forward(self, data):
        # 1. On déballe les données
        x, edge_index = data.x, data.edge_index 

        # 2. Première couche + Activation
        x = self.conv1(x, edge_index) # <--- On passe TOUJOURS edge_index
        x = x.relu()
        
        # 3. Deuxième couche
        x = self.conv2(x, edge_index)
        
        return x
```
